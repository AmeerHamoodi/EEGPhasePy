```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal
from EEGPhasePy.estimators import PHASTIMATE
```

```python
fs = 500

time_data = np.arange(0, 90, 1/fs)

clean_signal = np.sin(2 * np.pi * 10 * time_data)
ground_truth_phase = np.angle(signal.hilbert(clean_signal), deg=True) % 360

minimally_noisy_alpha = clean_signal + \
np.random.normal(0, 1, size=len(time_data))
```

```python
rt_filter = signal.firwin(80, [8, 12], fs=fs, pass_zero=False)
gt_filter = signal.firwin(300, [8, 12], fs=fs, pass_zero=False)

estimator = PHASTIMATE(rt_filter, gt_filter, fs)
estimator.optimize_parameters(minimally_noisy_alpha, method="genetic")
```
