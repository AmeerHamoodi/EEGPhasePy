```python
import numpy as np
import matplotlib.pyplot as plt

import scipy.signal as signal
from EEGPhasePy.estimators.phastimate import PHASTIMATE
```

```python
fs = 500

fir_filter = signal.firwin(128, [8, 12], fs=fs, pass_zero=False)

time_data = np.arange(0, 2, 1/fs)
signal_data = np.sin(
        2 * np.pi * 10 * time_data) + np.random.normal(0, 1, size=len(time_data))
filtered_signal = signal.filtfilt(fir_filter, 1.0, signal_data)
plt.plot(np.angle(signal.hilbert(filtered_signal[-250:]), deg=True))
```

```python
fir_filter = signal.firwin(80, [8, 12], fs=fs, pass_zero=False)

phastimate = PHASTIMATE(fir_filter, fir_filter, 500)
phastimate.predict(signal_data[-470:-220], 118)
```
